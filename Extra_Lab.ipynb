```python
import numpy as np
import pandas as pd

from google.colab import drive
drive.mount('/content/drive')
import os

# List everything in your main Google Drive folder
os.listdir('/content/drive/MyDrive')


url = "iris.csv"
df = pd.read_csv(url)

print("First 5 rows:\n", df.head(5))[cite: 1]
print("\nLast 3 rows:\n", df.tail(3))[cite: 1]
print("\nShape:", df.shape)[cite: 1]
print("\nSummary Info:")
df.info()[cite: 1]
print("\nDescriptive Statistics:\n", df.describe())[cite: 1]


print("\nRows 0, 4, 7, 8:\n", df.iloc[[0, 4, 7, 8]])[cite: 1]

# Slicing ranges (rows 3 to 7, rows 4 to 8 with cols 2 to 4)
slice_1 = df.iloc[3:8][cite: 1]
slice_2 = df.iloc[4:9, 2:5][cite: 1]
slice_3 = df.iloc[:, 1:4][cite: 1]


df_dropped = df.drop(index=4).drop(df.columns[3], axis=1)[cite: 1]


print("\nAverage Sepal Length:", df["sepal_length"].mean())[cite: 1]
print("Total Petal Length (Sum):", df["petal_length"].sum())[cite: 1]
print("Minimum Sepal Width:", df["sepal_width"].min())[cite: 1]
print("Maximum Petal Width:", df["petal_width"].max())[cite: 1]


print("\nMean attributes by Species:\n", df.groupby("species").mean())


X = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].to_numpy()


print("\nColumn Means (axis=0):", np.mean(X, axis=0))[cite: 3]
print("Row Sums - First 5 samples (axis=1):", np.sum(X, axis=1)[:5])[cite: 3]
print("Overall Max & Min:", np.max(X), np.min(X))[cite: 3]
print("Position of Global Max (argmax):", np.unravel_index(np.argmax(X), X.shape))[cite: 3]




sepal_len_corrected = X[:, 0] + 2.0[cite: 3]
sepal_len_f = (9 / 5) * X[:, 0] + 32[cite: 3]



mask = X[:, 0] > 6.0[cite: 3]
large_sepals = X[mask][cite: 3]
count_large_sepals = np.sum(mask)[cite: 3]
print(f"\nCount of samples with Sepal Length > 6.0: {count_large_sepals}")[cite: 3]



filtered_df = df[(df["petal_length"] > 5.0) & (df["species"] == "virginica")][cite: 1]



df_sorted = df.sort_values(by="sepal_length", ascending=False)[cite: 1]



conditions = [
    df["petal_length"] >= 5.0,
    (df["petal_length"] >= 2.0) & (df["petal_length"] < 5.0)
]
choices = ["Large", "Medium"]
df["Petal_Category"] = np.select(conditions, choices, default="Small")[cite: 1]


df["Sepal_Taxed"] = df["sepal_length"] * 0.90[cite: 1]



print("\nMissing values:\n", df.isnull().sum())[cite: 1]
df.rename(columns={"species": "Species_Class"}, inplace=True)[cite: 1]



reversed_x = X[::-1][cite: 2]
reshaped_X = X[:12, :].reshape(4, 3)[cite: 2, 3]
resized_X = np.resize(X, (2, 3))[cite: 2]



flat_arr = reshaped_X.flatten()[cite: 3]  # Deep copy
rav_arr = reshaped_X.ravel()   [cite: 3]  # View

rav_arr[0] = 999.0  # Alters reshaped_X[cite: 3]
flat_arr[1] = 888.0 # Leaves reshaped_X unaffected[cite: 3]

# Array Properties
print(f"\nArray Props -> Shape: {reshaped_X.shape}, Dim: {reshaped_X.ndim}, Size: {reshaped_X.size}, Dtype: {reshaped_X.dtype}")[cite: 2, 3]

# Unique values and frequencies
vals, counts = np.unique(X[:, 2], return_counts=True)[cite: 2]
most_frequent = vals[np.argmax(counts)][cite: 2]
freq_indices = np.where(X[:, 2] == most_frequent)[0][cite: 2]

# ==========================================
# 7. MATRIX ALGEBRA / OLS REGRESSION (Assignment 6 - Q4)[cite: 3]
# ==========================================
# Predict Petal Width (y) using Sepal Length, Sepal Width, and Petal Length (X_mat)
X_mat = X[:, :3]
y_vec = X[:, 3]

# Ordinary Least Squares: Beta = (X^T * X)^(-1) * X^T * y
XT = X_mat.T[cite: 3]
XTX_inv = np.linalg.inv(XT @ X_mat)[cite: 3]
beta = XTX_inv @ XT @ y_vec[cite: 3]

print("\nOrdinary Least Squares Regression Coefficients:", beta)[cite: 3]

# Predict for a custom sample [sepal_length=5.1, sepal_width=3.5, petal_length=1.4]
sample_input = np.array([5.1, 3.5, 1.4])
predicted_petal_width = sample_input @ beta[cite: 3]
print("Predicted Petal Width for sample:", round(predicted_petal_width, 3))[cite: 3]

# ==========================================
# 8. EXPORTING RESULTS (Assignment 3 - Q6l)[cite: 1]
# ==========================================
df.to_csv("processed_iris.csv", index=False)[cite: 1]
print("\nSuccessfully exported modified dataset to 'processed_iris.csv'")[cite: 1]
```
